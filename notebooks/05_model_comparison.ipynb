```python
import pandas as pd

results = pd.DataFrame({
    "Model": [
        "Multinomial Naive Bayes",
        "Random Forest"
    ],
    "Accuracy": [
        0.8007,
        0.8244
    ],
    "Macro F1": [
        0.67,
        0.78
    ],
    "Feature Request Recall": [
        0.19,
        0.55
    ]
})

results
```

```python
import matplotlib.pyplot as plt

results.set_index("Model").plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("FORCORA - ML Model Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend()
plt.show()
```
